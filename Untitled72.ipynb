```python
import pennylane as qml
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
import seaborn as sns
import time

from sklearn.datasets import make_moons, load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.decomposition import PCA
from sklearn.metrics import accuracy_score, balanced_accuracy_score
from sklearn.metrics import f1_score, confusion_matrix, roc_auc_score, roc_curve

np.random.seed(42)
torch.manual_seed(42)

print("Libraries imported successfully")
```

```python
N_QUBITS = 2
N_LAYERS = 3
BATCH_SIZE = 16
CIRCUIT_BUDGET = 24000
LEARNING_RATE = 0.025
SHOTS = None

print("Number of Qubits:", N_QUBITS)
print("Number of Layers:", N_LAYERS)
print("Circuit Budget:", CIRCUIT_BUDGET)
print("Learning Rate:", LEARNING_RATE)
```

```python
X_moons, y_moons = make_moons(
    n_samples=400,
    noise=0.20,
    random_state=42
)

plt.figure(figsize=(7,5))
plt.scatter(
    X_moons[:,0],
    X_moons[:,1],
    c=y_moons,
    cmap="coolwarm"
)

plt.title("Two Moons Dataset")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()

print("Dataset shape:", X_moons.shape)
```

```python
cancer = load_breast_cancer()

X_breast = cancer.data
y_breast = cancer.target

print("Dataset: Breast Cancer Wisconsin Diagnostic")
print("Total samples:", len(y_breast))
print("Total features:", X_breast.shape[1])
print("Classes:", cancer.target_names)
```

```python
def prepare_dataset(X, y):

    X_train, X_test, y_train, y_test = train_test_split(
        X, y,
        test_size=0.25,
        random_state=42,
        stratify=y
    )

    scaler = StandardScaler()

    X_train = scaler.fit_transform(X_train)
    X_test = scaler.transform(X_test)

    if X_train.shape[1] > 2:

        pca = PCA(n_components=2)
        X_train = pca.fit_transform(X_train)
        X_test = pca.transform(X_test)

    angle = MinMaxScaler(feature_range=(-np.pi, np.pi))

    X_train = angle.fit_transform(X_train)
    X_test = angle.transform(X_test)

    return (
        torch.tensor(X_train, dtype=torch.float32),
        torch.tensor(X_test, dtype=torch.float32),
        torch.tensor(y_train, dtype=torch.float32),
        torch.tensor(y_test, dtype=torch.float32)
    )


datasets = {
    "Two Moons": prepare_dataset(X_moons, y_moons),
    "Breast Cancer": prepare_dataset(X_breast, y_breast)
}

print("Preprocessing completed")
```

```python
dev = qml.device(
    "default.qubit",
    wires=N_QUBITS,
    shots=SHOTS
)

print("Quantum device created")
print(dev)
```

```python
@qml.qnode(dev, interface="torch",
           diff_method="parameter-shift")

def quantum_circuit(x, w0, w1, w2):

    qml.RY(x[0], wires=0)
    qml.RY(x[1], wires=1)

    weights = [w0, w1, w2]

    for layer in range(N_LAYERS):

        for wire in range(N_QUBITS):

            qml.Rot(
                *weights[layer][wire],
                wires=wire
            )

        qml.CNOT(wires=[0,1])

    return (
        qml.expval(qml.PauliZ(0)),
        qml.expval(qml.PauliZ(1))
    )


print("Quantum circuit created successfully")
```

```python
class QuantumNeuralNetwork(nn.Module):

    def __init__(self):

        super().__init__()

        self.weights = nn.ParameterList([
            nn.Parameter(
                0.1 * torch.randn(N_QUBITS, 3)
            )
            for _ in range(N_LAYERS)
        ])

    def forward(self, X):

        outputs = []

        for x in X:

            result = quantum_circuit(
                x,
                *list(self.weights)
            )

            value = torch.stack(result).mean()

            outputs.append(value)

        return 2.0 * torch.stack(outputs)


def set_trainable_layers(model, active):

    for i, p in enumerate(model.weights):

        p.requires_grad = i in active


print("QNN model created")
```

```python
def evaluate(model, X_test, y_test):

    model.eval()

    with torch.no_grad():

        logits = model(X_test).numpy()

    probabilities = 1 / (1 + np.exp(-logits))

    predictions = (probabilities >= 0.5).astype(int)

    actual = y_test.numpy().astype(int)

    return {
        "Accuracy": accuracy_score(actual, predictions),

        "Balanced Accuracy": balanced_accuracy_score(
            actual, predictions
        ),

        "Macro F1": f1_score(
            actual, predictions, average="macro"
        ),

        "AUROC": roc_auc_score(actual, probabilities),

        "Confusion Matrix": confusion_matrix(
            actual, predictions
        ),

        "Actual": actual,
        "Predictions": predictions,
        "Probabilities": probabilities
    }


print("Evaluation functions ready")
```

```python
def train_baseline(X_train, y_train):

    torch.manual_seed(100)

    model = QuantumNeuralNetwork()

    set_trainable_layers(model, [0, 1, 2])

    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=LEARNING_RATE
    )

    loss_fn = nn.BCEWithLogitsLoss()

    history = []
    gradients = []

    parameter_count = 18

    step_cost = BATCH_SIZE * (1 + 2 * parameter_count)

    steps = max(1, CIRCUIT_BUDGET // step_cost)

    start = time.time()

    for step in range(steps):

        indices = torch.randint(
            0, len(X_train), (BATCH_SIZE,)
        )

        xb = X_train[indices]
        yb = y_train[indices]

        optimizer.zero_grad()

        output = model(xb)

        loss = loss_fn(output, yb)

        loss.backward()

        grad = torch.cat([
            p.grad.flatten()
            for p in model.parameters()
            if p.grad is not None
        ])

        gradients.append(grad.norm().item())

        optimizer.step()

        history.append(loss.item())

    elapsed = time.time() - start

    return model, {
        "history": history,
        "gradients": gradients,
        "time": elapsed,
        "calls": steps * step_cost,
        "steps": steps
    }


print("Baseline training function ready")
```

```python
def train_progressive(X_train, y_train):

    torch.manual_seed(100)

    model = QuantumNeuralNetwork()

    history = []
    gradients = []

    total_calls = 0
    total_steps = 0

    loss_fn = nn.BCEWithLogitsLoss()

    stage_budget = CIRCUIT_BUDGET // N_LAYERS

    start = time.time()

    for layer in range(N_LAYERS):

        set_trainable_layers(model, [layer])

        optimizer = torch.optim.Adam(
            [model.weights[layer]],
            lr=LEARNING_RATE
        )

        parameter_count = 6

        step_cost = BATCH_SIZE * (
            1 + 2 * parameter_count
        )

        steps = max(1, stage_budget // step_cost)

        for step in range(steps):

            indices = torch.randint(
                0, len(X_train), (BATCH_SIZE,)
            )

            xb = X_train[indices]
            yb = y_train[indices]

            optimizer.zero_grad()

            output = model(xb)

            loss = loss_fn(output, yb)

            loss.backward()

            gradients.append(
                model.weights[layer].grad.norm().item()
            )

            optimizer.step()

            history.append(loss.item())

        total_calls += steps * step_cost
        total_steps += steps

    set_trainable_layers(model, [])

    elapsed = time.time() - start

    return model, {
        "history": history,
        "gradients": gradients,
        "time": elapsed,
        "calls": total_calls,
        "steps": total_steps
    }


print("Progressive training function ready")
```

```python
def run_experiment(name, data):

    X_train, X_test, y_train, y_test = data

    print("\nDataset:", name)

    baseline_model, baseline_info = train_baseline(
        X_train, y_train
    )

    progressive_model, progressive_info = train_progressive(
        X_train, y_train
    )

    baseline_metrics = evaluate(
        baseline_model, X_test, y_test
    )

    progressive_metrics = evaluate(
        progressive_model, X_test, y_test
    )

    return {
        "baseline_model": baseline_model,
        "progressive_model": progressive_model,
        "baseline_info": baseline_info,
        "progressive_info": progressive_info,
        "baseline_metrics": baseline_metrics,
        "progressive_metrics": progressive_metrics
    }


print("Comparison function ready")
```

```python
# FAST EXECUTION VERSION

BATCH_SIZE = 4
CIRCUIT_BUDGET = 300
LEARNING_RATE = 0.025

fast_datasets = {}

for name, data in datasets.items():

    X_train, X_test, y_train, y_test = data

    train_idx, _ = train_test_split(
        np.arange(len(X_train)),
        train_size=min(40, len(X_train)),
        random_state=42,
        stratify=y_train.numpy()
    )

    test_idx, _ = train_test_split(
        np.arange(len(X_test)),
        train_size=min(20, len(X_test)),
        random_state=42,
        stratify=y_test.numpy()
    )

    fast_datasets[name] = (
        X_train[train_idx],
        X_test[test_idx],
        y_train[train_idx],
        y_test[test_idx]
    )

print("Fast datasets prepared!")

results = {}

for name, data in fast_datasets.items():

    print("\nRunning:", name, flush=True)

    results[name] = run_experiment(name, data)

    print(name, "completed!", flush=True)

print("\nALL EXPERIMENTS COMPLETED SUCCESSFULLY!")
```

```python
rows = []

for name, result in results.items():

    for method in ["baseline", "progressive"]:

        metrics = result[method + "_metrics"]

        info = result[method + "_info"]

        rows.append({

            "Dataset": name,

            "Method": method,

            "Accuracy": metrics["Accuracy"],

            "Balanced Accuracy": metrics["Balanced Accuracy"],

            "Macro F1": metrics["Macro F1"],

            "AUROC": metrics["AUROC"],

            "Final Loss": info["history"][-1],

            "Gradient Variance": np.var(info["gradients"]),

            "Circuit Evaluations": info["calls"],

            "Training Time": info["time"]
        })


results_df = pd.DataFrame(rows)

display(results_df.round(4))

results_df.to_csv(
    "Quantum_Project_Results.csv",
    index=False
)

print("Results saved successfully")
```
